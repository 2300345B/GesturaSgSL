```python
import cv2

print("🔍 Scanning for available webcams...")
for i in range(10):  # Try camera indexes 0–9
    cap = cv2.VideoCapture(i)
    if cap.read()[0]:
        print(f"✅ Camera found at index {i}")
        cap.release()
    else:
        print(f"❌ No camera at index {i}")
```
