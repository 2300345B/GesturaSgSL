```python
import subprocess
import os
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

# Configuration
img_size = 800
batch_size = 16
epochs = 100
dataset_yaml = os.path.abspath("dataset.yaml")
weights = os.path.abspath("yolov5m.pt")
model_name = "sgsl_model_improved"

print("🚀 Starting YOLOv5 training...\n")

# Run YOLOv5 training with live output streaming
try:
    process = subprocess.Popen(
        [
            "python", "yolov5/train.py",
            "--img", str(img_size),
            "--batch", str(batch_size),
            "--epochs", str(epochs),
            "--data", dataset_yaml,
            "--weights", weights,
            "--name", model_name,
            "--cache"
        ],
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True
    )

    # Print training output live
    for line in process.stdout:
        print(line, end='')

    process.wait()

    if process.returncode != 0:
        raise subprocess.CalledProcessError(process.returncode, process.args)

except subprocess.CalledProcessError as e:
    print("\n❌ Training failed!")
    print(e)
    exit(1)

# Path to training results
run_folder = f"runs/train/{model_name}"
print(f"\n✅ Training complete. Check the folder: {run_folder}/")
```

```python
results_img_path=os.path.join(run_folder,'results.png')
if os.path.exists(results_img_path):
    img=Image.open(results_img_path)
    plt.figure(figsize=(10,5))
    plt.imshow(img)
    plt.axis('off')
    plt.title('YOLOv5 Training Metrics')
    plt.show()

else:
    print('results.png not found')
```

```python
csv_path=os.path.join(run_folder,'results.csv')
if os.path.exists(csv_path):
    df=pd.read_csv(csv_path)
    df.columns=df.columns.str.strip()

    plt.figure(figsize=(12,6))
    plt.plot(df['train/box_loss'],label='Train Box Loss')
    plt.plot(df['val/box_loss'],label='Val Box Loss')
    plt.plot(df['train/obj_loss'],label='Train Obj Loss')
    plt.plot(df['val/obj_loss'],label='Val Box Loss')
    plt.plot(df['metrics/mAP_0.5'],label='mAP@0.5')
    plt.xlabel("Epochs")
    plt.ylabel("Loss/mAP")
    plt.title
   
```
