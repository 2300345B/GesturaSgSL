```python
import cv2
import os
import time
import uuid

# Path where images will be saved
images_path = 'dataset/images/train2'

# List of labels for which images will be collected
labels = [
    'yes', 'no','me','you','hello', 'hi', 'good',
    'ok', 'sorry', 'i love you', 'thank you', 'welcome'
]

# Number of images to collect per label
number_imgs = 100

# Create the main images directory if it doesn't exist
if not os.path.exists(images_path):
    os.makedirs(images_path)

for label in labels:
    print(f'\nCollecting images for "{label}"')
    label_path = os.path.join(images_path, label)
    os.makedirs(label_path, exist_ok=True)

    # Give time to prepare
    print(f'You have 8 seconds to prepare for the gesture: "{label}"')
    time.sleep(8)

    cap = cv2.VideoCapture(0)
    if not cap.isOpened():
        print("Error: Could not open camera.")
        continue
    print(f'Starting capture for "{label}"...\n')
    time.sleep(2)  # camera warmup

    for imgnum in range(number_imgs):
        ret, frame = cap.read()
        if not ret:
            print("Error: Failed to capture frame.")
            break

        imgname = os.path.join(label_path, f'{label}.{uuid.uuid1()}.jpg')
        cv2.imwrite(imgname, frame)
        print(f'  [{imgnum + 1}/{number_imgs}] Captured image for "{label}"')
        cv2.imshow('Collecting Images - press Q to quit', frame)

        time.sleep(0.4)  # Faster capture (2.5 fps)

        if cv2.waitKey(1) & 0xFF == ord('q'):
            print("  Manually stopped.")
            break

    cap.release()
    time.sleep(2)  # short rest between labels

cv2.destroyAllWindows()
print(f'\n✅ Image collection complete. Images saved in: {images_path}')
# End of script
```
